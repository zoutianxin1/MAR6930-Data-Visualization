## 1. Setup and data

Keep this notebook and `tickets.csv` in the same folder. Select a Python kernel. If a package is missing, uncomment the installation line below and run it once.

The data are copied from the Lecture 1 course materials, with all values unchanged. Each row is one month; `tickets_received` and `tickets_processed` are counts. The `employees` column is not used in this chart.

```python
# %pip install pandas matplotlib

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FormatStrFormatter

plt.style.use("default")
tickets = pd.read_csv("tickets.csv")
months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December",
]

tickets
```

## 2. Original cluttered example

Run the code below to draw the starting chart.

```python
fig, ax = plt.subplots(figsize=(9, 5))

# Draw both lines with a marker at every month.
ax.plot(months, tickets["tickets_received"], marker="D",
        label="Ticket Volume Received")
ax.plot(months, tickets["tickets_processed"], marker="s",
        label="Ticket Volume Processed")

# Keep the decimal tick labels, angled month names, and gridlines.
ax.set_ylim(0, 300)
ax.set_yticks(range(0, 301, 50))
ax.yaxis.set_major_formatter(FormatStrFormatter("%.2f"))
plt.xticks(rotation=45, ha="right")
ax.grid(axis="y")

# Keep the plot border and a separate legend below the chart.
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.3), ncol=2)

plt.tight_layout()
plt.show()
```
